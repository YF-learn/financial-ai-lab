```python
class Point:
    __match_args__ = ('x', 'y')
    def __init__(self, x, y):
        self.x = x
        self.y = y
points=[Point(1, 2),Point(0, 3)]


match points:
    case []:
        print("No points")
    case [Point(0, 0)]:
        print("The origin")
    case [Point(x, y)]:
        print(f"Single point {x}, {y}")
    case [Point(0, y1), Point(0, y2)]:
        print(f"Two on the Y axis at {y1}, {y2}")
    case _:
        print("Something else")
```

### define function 

```python
def fib(n):
    '''打印小于n的斐波那契数列'''
    # 没有返回值,只能打印
    a,b=0,1
    while a<n:
        print(a,end=' ')
    a,b=b,a+b

fib(100)
```

```python
# 编写一个返回斐波那契数列的数组
def fib2(n):
    '''返回小于n的斐波那契数列'''
    result=[]
    a,b=0,1
    while a<n:
        result.append(a)
        a,b=b,a+b
    return result   #return 语句不带表达式参数时，返回 None。函数执行完毕退出也返回 None

fib2(200)
```

```python
def ask_ok(prompt,retries=3,reminder="please try again!"):
    while True:
        reply=input(prompt)
        if reply in {'y','yes','ye','yep'}:
            return True
        if reply in {'n','no','nop','nope'}:
            return False
        retries-=1
        if retries<0:
            raise ValueError(reminder)
        print(reminder)

# ask_ok("do you really want to quit?")
ask_ok('are you serious?',2,'try again!')

       
```

❗️默认值为可变对象是，会共享内存，导致修改默认值会影响所有调用

```python
def f(a,L=[]):
    L.append(a)
    return L

print(f(1))
print(f(2))
print(f(3))
```

```python
# 不想在后续调用之间共享默认值时，应以如下方式编写函数
def f (a,L=None):
    # 如果L为None，创建一个新的空列表
    if  L is None:
        L=[]
    L.append(a)
    return L

print(f(1))
print(f(2))
print(f(3))
```

-- (2) 关键字参数

```python
def parrot(voltage, state='a stiff', action='voom', type='Norwegian Blue'):
    print("-- This parrot wouldn't", action, end=' ')
    print("if you put", voltage, "volts through it.")
    print("-- Lovely plumage, the", type)
    print("-- It's", state, "!")
```

```python
# 函数调用时，关键字参数必须跟在位置参数后面。所有传递的关键字参数都必须匹配一个
# 函数接受的参数（比如，actor 不是函数 parrot 的有效参数），关键字参数的顺序并不重要
parrot(1000)                                          # 1 个位置参数
parrot(voltage=1000)                                  # 1 个关键字参数
parrot(voltage=1000000, action='VOOOOOM')             # 2 个关键字参数
parrot(action='VOOOOOM', voltage=1000000)             # 2 个关键字参数
parrot('a million', 'bereft of life', 'jump')         # 3 个位置参数
parrot('a thousand', state='pushing up the daisies')  # 1 个位置参数，1 个关键字参数
```

```python
# 以下调用无效
parrot()                     # 缺失必需的参数
parrot(voltage=5.0, 'dead')  # 关键字参数后存在非关键字参数
parrot(110, voltage=220)     # 同一个参数重复的值
parrot(actor='John Cleese')  # 未知的关键字参数
```

```python
def cheeseshop(kind, *arguments, **keywords): # *name 形参接收一个 元组，**name 形参接收一个 字典，
    print("-- Do you have any", kind, "?")
    print("-- I'm sorry, we're all out of", kind)
    for arg in arguments:
        print(arg)
    print("-" * 40)
    for kw in keywords:
        print(kw,":",keywords[kw])
```

- 特殊参数

1.标准参数

2.仅限位置参数

3.仅限关键字参数

4.混合使用

```python
# 标准参数
def f1(arg):
    print(arg)

# 仅限位置参数
def f2(arg,/):
    print(arg)

# 仅限关键字参数
def f3(*,arg):
    print(arg)

# 混合使用
def f4(pos_only,/,standard,*,kwd_only):
    print(pos_only,standard,kwd_only)
```

```python
f1(2)
f1(arg=2)
f2(arg=2) # 报错，f2() 仅限位置参数
```

```python
f3(2) # 报错，f3() 仅限关键字参数
```

```python
f4(1,2,3)
```

```python
f4(1,2,kwd_only=3)
```

```python
f4(1,standard=2,kwd_only=3) # 正确
```

```python
f4(pos_only=1,standard=2,kwd_only=3) # 报错，pos_only 仅限位置参数
```

仅限位置形参的名称可以在 **kwds 中使用，而不产生歧义

```python
# 因为关键字 'name' 总与第一个形参绑定，调用时会报错
def foo(name,**kwds):
    return 'name' in kwds

foo(1,**{'name':2})
```

```python
# 确认是name是位置参数后，name可以作为关键字参数使用
def foo(name,/,**kwds):
    return 'name' in kwds

foo(1,**{'name':2})
```

用法一：sorted()/list.sort()自定义排序

```python
# 按绝对值排序
nums = [-3, 1, -5, 2]
sorted(nums, key=lambda x: abs(x))
# [1, 2, -3, -5]

# # 按字符串长度排序
words = ['apple', 'hi', 'banana']
sorted(words, key=lambda w: len(w))
# ['hi', 'apple', 'banana']

# # 按字典的某个值排序
users = [{'name': 'Bob', 'age': 25}, {'name': 'Alice', 'age': 30}]
sorted(users, key=lambda u: u['age'])
# [{'name': 'Bob', 'age': 25}, {'name': 'Alice', 'age': 30}]


# max()/min() 函数也可以接收一个关键字参数 key 来指定排序的规则
max(nums, key=abs)
# 5
```

语法结构：

    参数2: 注解值 = 默认值  # 带默认值的参数也能加注解

    """函数文档字符串"""
