### 7.1.1 格式化字符串面值

```python
import math
f'the value of pi is approximately {math.pi:.2f}'
```

```python
#  ':' 后传递整数，为该字段设置最小字符宽度，常用于列对齐：
table = {'Sjoerd': 4127, 'Jack': 4098, 'Dcab': 7678}
for name,phone in table.items():
    print(f'{name:10} ==> {phone:7d}')
```

```python
# '!a' 应用 ascii() ，'!s' 应用 str()，'!r' 应用 repr(),格式化前转换值
animals = 'eels'
print(f'My hovercraft is full of {animals}.')
print(f'My hovercraft is full of {animals!r}.')
```

```python
# = 说明符可被用于将一个表达式扩展为表达式文本、等号再加表达式求值结果的形式
bugs = 'roaches'
count = 13
area = 'living room'
print(f'Debugging {bugs=} {count=} {area=}')
print(f'Debugging {bugs} {count} {area}')
```

### 7.1.2 字符串format方法

```python
    print('We are the {} who say "{}!"'.format('knights', 'Ni'))
```

```python
print('I like {1} and {0}'.format('apple','pear'))
```

```python
# 位置参数和关键字参数可以任意组合
print('The story of {0}, {1}, and {other}.'.format('Bill', 'Manfred',
                                                   other='Georg'))
```

```python
# 不拆分长字符串
table = {'Sjoerd': 4127, 'Jack': 4098, 'Dcab': 8637678}
print('Jack: {0[Jack]:d}; Sjoerd: {0[Sjoerd]:d}; '
      'Dcab: {0[Dcab]:d}'.format(table))
```

```python
print('Jack: {Jack:d}; Sjoerd: {Sjoerd:d}; '
      'Dcab: {Dcab:d}'.format(**table))
```

```python
table = {k: str(v) for k, v in vars().items()}
message = " ".join([f'{k}: ' + '{' + k +'};' for k in table.keys()])
print(message.format(**table))
```

```python
for x in range (1,11):
    print('{0:2d} {1:3d} {2:4d} '.format(x,x**2,x**3))
```

### 7.1.3 手动格式化字符串


```python
for x in range(1, 11):
    print(repr(x).rjust(2), repr(x*x).rjust(3), end=' ') # rjust() 方法用于右对齐字符串
    # 请注意上一行中 'end' 的使用
    print(repr(x*x*x).rjust(4))
```

```python
# str.zfill() ，该方法在数字字符串左边填充零，且能识别正负号：
'12'.zfill(5) # 返回'00012'

'-3.14'.zfill(7) # 返回'-003.14'

'3.14159265359'.zfill(5)
```

### 7.2.2 使用json保存结构化数据

- 4个核心函数区分，带s的是转字符串
1. json.jump(obj,file):把python对象写入json文件
2. json.jumps(obj):把python对象转换为json字符串
3. json.load(file):从json文件读取python对象
4. json.loads(str):从json字符串读取python对象

```python
import json
from datetime import datetime
import numpy as np

# 模拟金融场景的嵌套结构化数据（股票基本信息+日线+指标）
stock_data = {
    "code": "000001.SZ",
    "name": "平安银行",
    "industry": "银行",
    "is_st": False,
    "daily_kline": [  # 嵌套列表
        {
            "date": datetime(2024, 1, 2).strftime('%Y-%m-%d'),  # JSON不认datetime，要转字符串！
            "open": 10.25,
            "close": 10.32,
            "volume": 1234567
        },
        {
            "date": datetime(2024, 1, 3).strftime('%Y-%m-%d'),
            "open": 10.32,
            "close": 10.45,
            "volume": 1456789
        }
    ],
    "indicator": {  # 嵌套字典
        "pe": 5.23,
        "pb": 0.51,
        "roe": 10.2
    }
}

# 【写入JSON文件】—— 必配3个参数！
with open('stock_000001.json', 'w', encoding='utf-8') as f:
    json.dump(
        stock_data, f,
        ensure_ascii=False,  # ✅ 保留中文（不加的话会把"平安银行"转成\u5e73\u5b89\u94f6\u884c）
        indent=2,            # ✅ 格式化缩进2空格（方便人看，金融数据要可读性）
        sort_keys=True       # ✅ 按键名排序（方便对比不同时间的股票数据）
    )

# 【读回JSON文件】
with open('stock_000001.json', 'r', encoding='utf-8') as f:
    loaded_data = json.load(f)

# 验证
print(loaded_data['name'])              # → 平安银行
print(loaded_data['daily_kline'][0]['date'])  # → 2024-01-02
print(loaded_data['indicator']['pe'])   # → 5.23
```

JSON 只认 Python基础类型（dict/list/str/int/float/bool/None），遇到 datetime、numpy数组 会报错，有两种解决方式：
1. 手动转换
2. 用default参数自动转换
3. 与pandas配合，df.to_json():将数据帧转化为json,pd.read_json()：读取json格式数据
