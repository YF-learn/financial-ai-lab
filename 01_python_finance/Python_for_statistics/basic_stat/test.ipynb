```python
# 类
class User:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def say(self):
        print(f'我是{self.name}，我今年{self.age}岁')
    
    def age_next_year(self):
        print(f'我明年{self.age+1}岁')


user1=User(name='zyf',age=18)
user1.say()
user1.age_next_year()
```

```python
import numpy as np
import pandas as pd
```

```python
array1=np.array([1,2,3,4,5])
array1
```

```python
array1+3,array1*2
```

```python
# 数组只能存放相同类型的数据，如果同时存放不同类型的的数据，会自动转换为相同类型的数组
array1=np.array([1,2,3,4,'hello'])
array1


```

```python
# 多维数组
array2=np.array([[1,2,3],[4,5,6]])
array2
```

```python
# 获取数组的形状和维度
array1.shape # (5,)
array1.ndim # 1
array2.shape # (2,3)
# array2.ndim  # 2
```

```python
# 生成等差数列的方法，arrange函数 arrange(start,stop,step)
a1=np.arange(10)
a2=np.arange(1,10)
a3=np.arange(1,10,2) 

a1,a2,a3

```

```python
# 多种生成数组的方法
# np.tile函数
a1=np.tile(a1,2)
a2=np.tile(a2,(2,1))
a3=np.tile(a3,(2,1))
a1,a2,a3


```

```python
zeros=np.zeros((2,3))
ones=np.ones((2,3))
np.tile(0,3)
zeros,ones
```

```python
# 数组的切片
a1[0:5]
a2[:,0:2]
```

```python
# dataframe
l=['A','B','C','D','E','F','G','H','I','J']
df1=pd.DataFrame({'col1':range(10),'col2':l})
df1

```

```python
df2=pd.read_csv("../pystat-code-2021-01-25/2-4-1-sample_data.csv")
df2
```

```python
# 纵向连接dataframe
df3=pd.concat([df1,df2],axis=0)
df3
# 横向连接dataframe
df4=pd.concat([df1,df2],axis=1)
df4
```

```python
df3['col2']
# 或者
df3.col2

df4[['col1','col2']]

```

```python
# 删除列
df4.drop(columns=['col2'],inplace=True)
df4
```

```python
# head()函数查看前n行
df3.head(3)
# tail()函数查看后n行
df3.tail(3)
df3
```

```python
# query()函数根据条件筛选数据,最后指定返回的列名
df3.query('col2=="B" | col2=="C"')['col1'] # &、|、~分别表示与、或、非操作
```
